**Introduction**

Exploratory Data Analysis (EDA) is the process of examining, cleaning, summarizing, and visualizing data to identify meaningful patterns, trends, and relationships.

Our dataset contains reported crime incidents in Los Angeles from 2020 to 2024, along with information about the crime type, time, location, and victim characteristics.

**Dataset Information**

Our dataset contains reported crime incidents within the City of Los Angeles from 2020 to 2024. The incidents are categorized across the 21 geographic areas of the Los Angeles Police Department (LAPD). It includes information related to crime type, date and time of occurrence, location, victim details, and other crime-related attributes. The purpose of this analysis is to identify crime patterns based on crime type, time, location, and victim characteristics.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
```

```python
df = pd.read_csv("crime_datasets.csv")

print("Dataset Loaded Successfully")

print("Shape:", df.shape)

display(df.head())

print("\nDataset Information:")

df.info()

print("\nColumn Names:")

print(df.columns.tolist())
```

**Data Cleaning**

Data cleaning is the process of identifying and correcting missing, duplicate, incorrect, inconsistent, or invalid data to improve the quality, accuracy, and reliability of the dataset before analysis.


**Data Cleaning Steps**

1.  Handle Missing Values – Identify and handle missing data using suitable imputation methods.
2.  Remove Duplicate Records – Identify and remove duplicate rows from the dataset.


1.  Correct Data Types – Convert features into their appropriate data types.
2.  Handle Invalid Values – Identify and correct unrealistic or inconsistent values.

  
1. Detect Outliers – Identify unusual values using suitable outlier detection methods.  
2. Extract Useful Features – Derive features such as year, month, hour, and time of day.


1.  Validate Cleaned Data – Recheck the dataset to ensure consistency and completeness.



**Data Cleaning Techniques Used**

1.  Handle Missing Values – Median Imputation for numerical data and Mode Imputation for categorical data.
2.  Remove Duplicate Records – Duplicate Record Removal using drop_duplicates().

1.   Correct Data Types – Data Type Conversion using astype() and pd.to_datetime().
2.   Handle Invalid Values – Rule-Based Filtering and Replacement of Invalid Values.

1.   Detect Outliers – Interquartile Range (IQR) Method.
2.   Extract Useful Features – Date-Time Feature Extraction for Year, Month, Hour, and Time of Day.


1.   Validate Cleaned Data – Data Validation using null-value, duplicate, data-type, and range checks.












```python
# Check missing values
print("Missing Values Before Cleaning:")
print(df.isnull().sum())

# Remove duplicate records
df = df.drop_duplicates()

# Convert date columns
df['DATE OCC'] = pd.to_datetime(df['DATE OCC'], errors='coerce')
df['Date Rptd'] = pd.to_datetime(df['Date Rptd'], errors='coerce')

# Handle missing categorical values using Mode Imputation
categorical_columns = df.select_dtypes(include='object').columns

for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])

# Handle missing numerical values using Median Imputation
numerical_columns = df.select_dtypes(include=np.number).columns

for column in numerical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].median())

# Handle invalid victim ages
df.loc[(df['Vict Age'] <= 0) | (df['Vict Age'] > 100), 'Vict Age'] = np.nan
df['Vict Age'] = df['Vict Age'].fillna(df['Vict Age'].median())

# Outlier detection using IQR method
Q1 = df['Vict Age'].quantile(0.25)
Q3 = df['Vict Age'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df['Vict Age'] < lower_limit) |
              (df['Vict Age'] > upper_limit)]

print("Number of Outliers:", len(outliers))

# Extract useful time features
df['Year'] = df['DATE OCC'].dt.year
df['Month'] = df['DATE OCC'].dt.month
df['Month Name'] = df['DATE OCC'].dt.month_name()

# Convert TIME OCC into hour
df['Hour'] = df['TIME OCC'] // 100

# Create time-of-day feature
def get_time_period(hour):
    if hour < 6:
        return 'Night'
    elif hour < 12:
        return 'Morning'
    elif hour < 18:
        return 'Afternoon'
    else:
        return 'Evening'

df['Time of Day'] = df['Hour'].apply(get_time_period)

print("\nCleaning Completed")
print("Final Shape:", df.shape)
print("Remaining Missing Values:", df.isnull().sum().sum())
print("Remaining Duplicate Rows:", df.duplicated().sum())
```

**Basic EDA Calculation Terms**

1.  Mean – The average value of a numerical feature.

1.  Median – The middle value when the data is arranged in order.

1.  Mode – The most frequently occurring value in the data.
2.  Minimum – The smallest value present in the data.

2.  Maximum – The largest value present in the data.

1.  Standard Deviation – Measures how much the values vary from the mean.
2.   Variance – Measures the overall spread of values around the mean.




```python
print("===== BASIC EDA STATISTICS =====")

numeric_data = df.select_dtypes(include=np.number)

mean_values = numeric_data.mean()
median_values = numeric_data.median()
mode_values = numeric_data.mode().iloc[0]
min_values = numeric_data.min()
max_values = numeric_data.max()
std_values = numeric_data.std()
variance_values = numeric_data.var()

statistics = pd.DataFrame({
    'Mean': mean_values,
    'Median': median_values,
    'Mode': mode_values,
    'Minimum': min_values,
    'Maximum': max_values,
    'Standard Deviation': std_values,
    'Variance': variance_values
})

pd.set_option('display.float_format', '{:,.2f}'.format)

display(statistics)
```

```python
# Graph 1: Horizontal Bar Chart

#Crime Type Distribution
crime_type = df['Crm Cd Desc'].value_counts().head(15).reset_index()

crime_type.columns = ['Crime Type', 'Number of Crimes']

fig = px.bar(
    crime_type,
    x='Number of Crimes',
    y='Crime Type',
    orientation='h',
    title='Top 15 Crime Types in Los Angeles - Horizontal Bar Chart',
    labels={
        'Number of Crimes': 'Number of Reported Crimes',
        'Crime Type': 'Crime Type'
    },
    hover_data=['Number of Crimes']
)

fig.update_layout(
    yaxis={'categoryorder': 'total ascending'},
    height=650
)

fig.show()
```

The distribution extends from around 142 to 193 crimes, showing noticeable variation between months.The monthly crime distribution is fairly stable with moderate consisderation. Morning has a considerably lower number of incidents, while Night has the lowest among the four periods. ```python
# Graph 10:  KDE Density Plot

#Time-Of-Day Distribution of Top Crime Types
import plotly.figure_factory as ff

top_crimes = df['Crm Cd Desc'].value_counts().head(5).index

ridge_data = df[
    df['Crm Cd Desc'].isin(top_crimes)
].copy()

ridge_data = ridge_data[
    (ridge_data['Hour'] >= 0) &
    (ridge_data['Hour'] <= 23)
]

hist_data = []
group_labels = []

for crime in top_crimes:
    values = ridge_data[
        ridge_data['Crm Cd Desc'] == crime
    ]['Hour'].dropna().values

    hist_data.append(values)
    group_labels.append(crime)

fig = ff.create_distplot(
    hist_data,
    group_labels,
    bin_size=1,
    show_hist=False,
    show_rug=False
)

fig.update_layout(
    title='Time-of-Day Distribution of Top Crime Types - KDE Density Curve Plot',
    xaxis_title='Hour of Day',
    yaxis_title='Density',
    height=650
)

fig.show()
```

**Inference - Time-of-Day Distribution of Top Crime Types**

The graph shows that different crimes happen at different times of the day. Theft of Identity has the sharpest peak, around midday (hour 12), and drops quickly after that. Theft Plain – Petty peaks in the early afternoon (around hour 13). Battery – Simple Assault peaks later, in the late afternoon (around hour 16-17). Vehicle – Stolen rises slowly through the day and peaks in the evening (around hour 18-19). Burglary stays almost flat at about 0.04 for most of the day, with a small rise in the evening. This indicates that crime timing varies by crime type: identity and petty thefts are mostly daytime crimes, while assault and vehicle theft are more common in the evening.patterns.

A few points stand out clearly. Vehicle – Stolen has the highest single point, close to 200 crimes, while Theft from Motor Vehicle – Petty and Burglary have smaller but still noticeable spikes. ```python
# Graph 12: Radar Chart

#Crime Profile of Top 5 Areas
top_areas = df['AREA NAME'].value_counts().head(5).index
top_crimes = df['Crm Cd Desc'].value_counts().head(6).index

radar_data = df[
    df['AREA NAME'].isin(top_areas) &
    df['Crm Cd Desc'].isin(top_crimes)
]

radar_table = pd.crosstab(
    radar_data['AREA NAME'],
    radar_data['Crm Cd Desc']
)

radar_table = radar_table.reindex(
    index=top_areas,
    columns=top_crimes,
    fill_value=0
)

fig = go.Figure()

for area in radar_table.index:
    values = radar_table.loc[area].tolist()
    values.append(values[0])

    categories = radar_table.columns.tolist()
    categories.append(categories[0])

    fig.add_trace(
        go.Scatterpolar(
            r=values,
            theta=categories,
            fill='toself',
            name=area
        )
    )

fig.update_layout(
    title='Crime Profile Comparison of Top 5 Areas - Radar Chart',
    polar=dict(
        radialaxis=dict(
            visible=True
        )
    ),
    height=700
)

fig.show()
```

**Inference - Crime Profile Comparison of Top 5 Areas**

The graph shows that each area has its own crime profile. Pacific has the largest shape overall, with the highest Burglary, high Battery – Simple Assault, and a noticeable Burglary from Vehicle count. Mission is the opposite: it has the highest Vehicle – Stolen (about 35) but is low in Burglary and Battery. Southeast is also strong in Vehicle – Stolen and Battery, while N Hollywood has moderate values across most categories, with Vehicle – Stolen and Burglary standing out. Olympic has a smaller shape, with Theft Plain – Petty as its main category. Theft of Identity is low in all five areas. This indicates that the type of crime depends strongly on the area, so each area may need a different focus for crime prevention.
